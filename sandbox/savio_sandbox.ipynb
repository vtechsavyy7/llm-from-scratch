```python
import tiktoken
import numpy as np
```

```python
# Get the encoding for the 'gpt2' model
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
# Try tokenizing some text

text = "It's understandable that you're overwhelmed with the staggering amount of information available in supercalifragilisticexpialidocious."

tokens = tokenizer.encode(text)

print(" Tokens are: ", tokens)
print(" Number of tokens: ", len(tokens))
```

```python
# Get the vocabulary size of the tokenizer
vocab_size = tokenizer.n_vocab
print(" Vocabulary size: ", vocab_size)
```

```python
# Increment each token by 1 to create a new list of tokens
# incremented_tokens = [token + 1 for token in tokens]

# Decode the incremented tokens back to text
decoded_text = tokenizer.decode(tokens)

print(" Decoded text from incremented tokens: ", decoded_text)
```

```python
# Decode the original tokens individually to see the corresponding text for each token
decoded_tokens = [tokenizer.decode([token]) for token in tokens]

print(" Decoded tokens individually: ", decoded_tokens)
```

```python
# Load the text file verdict.txt
with open("../data/the-verdict.txt", "r") as file:
    verdict_text = file.read()

print(verdict_text[:100])  # Print the first 100 characters of the verdict text
print(type(verdict_text))

# Encode the text using the GPT-2 tokenizer from tiktoken
enc_text = tokenizer.encode(verdict_text)

print(len(enc_text))
print(type(enc_text))
print(enc_text[:10])  # Print the first 100 encoded tokens
```

```python
# Create a sliding window sampler: Create inputs-target pairs for training
# The targets are one step shifted ahead of the input

window_size = 256  # Define the size of the sliding window

inputs = []
targets = []
for i in range(0, len(enc_text) - window_size):
    inputs.append(enc_text[i:i + window_size])
    targets.append(enc_text[i + 1:i + window_size + 1])

print(len(inputs))
print(len(targets))
print(inputs[0][:10])
print(targets[0][:10])
```
